```python
from statsmodels.stats.proportion import proportions_ztest

success = 70
total = 1000
old_rate = 0.05

z, p = proportions_ztest(success, total, old_rate, alternative='larger')

print("Z value:", z)
print("P-value:", p)

if p < 0.05:
    print("Reject H0")
    print("The conversion rate has significantly increased.")
else:
    print("Fail to reject H0")
    print("There is not enough evidence of an increase.")
```

```python
from scipy.stats import chi2_contingency

data = [
    [30, 20],
    [20, 30]
]

chi2, p, dof, expected = chi2_contingency(data)

print("Chi-Square value:", chi2)
print("P-value:", p)
print("Degrees of freedom:", dof)

print("Expected values:")
print(expected)

if p < 0.05:
    print("Reject H0")
    print("Gender and product preference are associated.")
else:
    print("Fail to reject H0")
    print("Gender and product preference are independent.")
```

```python
#A college wants to know whether students' department is related to their food preference.
from scipy.stats import chi2_contingency

data = [
    [40, 30, 30],   # CSE
    [30, 40, 30],   # ECE
    [20, 30, 50]    # MECH
]
chi2, p, dof, expected = chi2_contingency(data)

print("Chi-Square:", chi2)
print("P-value:", p)
print("Degrees of freedom:", dof)

if p < 0.05:
    print("Reject H0")
    print("Department and food preference are related.")
else:
    print("Fail to reject H0")
    print("Department and food preference are not related.")
```
