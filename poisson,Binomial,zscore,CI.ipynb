```python
sample_rating=[8.1,8.6,8.8,8.1,8,5,8.6,8.7,8.8,8.9,9.1,9.2,9.3,9.4,9.4,9.0,9.1,8.9,8.3,8.5,8.6,8.7,8.0,9.0,9.1,9.2,9.3,8.2,8.3,7.9,8.0,9.6]
mean=np.mean(sample_rating)
n=31
std=np.std(sample_rating)
```

```python
SE=stats.sem(np.random.normal(mean,std,n))
SE
```

```python
ans=stats.norm.interval(confidence=0.95,loc=mean,scale=SE)
ans
```
