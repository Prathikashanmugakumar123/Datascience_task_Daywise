```python
import numpy as np
from scipy import stats

# Training time of 10 employees
data = [42, 45, 38, 40, 44, 43, 41, 39, 46, 42]

# 95% Confidence Interval automatically
ci = stats.t.interval(
    confidence=0.95,
    df=len(data) - 1,
    loc=np.mean(data),
    scale=stats.sem(data)
)

print("95% Confidence Interval:")
print("Lower:", ci[0])
print("Upper:", ci[1])
```

```python
import numpy as np
from scipy import stats

# Training time of 30 employees
data = [
    42, 45, 38, 40, 44, 43, 41, 39, 46, 42,
    44, 40, 43, 45, 41, 42, 39, 44, 46, 40,
    43, 42, 45, 41, 44, 39, 43, 42, 40, 46
]

population_std = 3

ci = stats.norm.interval(
    0.95,
    loc=np.mean(data),
    scale=population_std / np.sqrt(len(data))
)

print("Number of data:", len(data))
print("Mean:", np.mean(data))

print("95% Confidence Interval:")
print("Lower:", ci[0])
print("Upper:", ci[1])
```
